```python
import matplotlib.pyplot as plt
```

##Ejercicio 1: Evolución de la temperatura (Gráfico de Líneas)

Crea un gráfico de líneas para mostrar el registro de temperatura durante una semana

```python
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
temperaturas = [22, 24, 19, 21, 25, 28, 27]

plt.plot(dias, temperaturas, marker="o")
plt.title("Temperatura durante la semana")
plt.xlabel("Día")
plt.ylabel("Temperatura °C")
plt.grid(True)
plt.show()
```

## Ejercicio 2: Ventas por tipo de fruta (Gráfico de Barras)

Haz un gráfico de barras para comparar las ventas semanales de diferentes frutas.


```python
frutas = ["Manzanas", "Plátanos", "Naranjas", "Fresas"]
kilos = [50, 80, 45, 30]

plt.bar(frutas, kilos, color="green")
plt.title("Ventas semanales por tipo de fruta")
plt.xlabel("Fruta")
plt.ylabel("Kilos vendidos")
plt.show()
```

## Ejercicio 3: Horas de estudio vs. Calificación (Gráfico de Dispersión)

Genera un gráfico de dispersión (scatter plot) para analizar si existe una relación entre las horas que estudian los
alumnos y la nota que obtienen.


```python
horas = [1, 2, 3, 4, 5, 6, 7, 8]
notas = [50, 55, 65, 70, 80, 85, 92, 98]

plt.scatter(horas, notas, color="green")
plt.title("Horas de estudio vs. Calificación")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.grid(True)
plt.show()
```

## Ejercicio 4: Distribución de mascotas favoritas (Gráfico de Pastel)

Utiliza un gráfico circular para mostrar el porcentaje de preferencia de mascotas en una clase. Tip: Usa
autopct="%1.1f%%".


```python
mascotas = ["Perros", "Gatos", "Peces", "Aves"]
cantidad = [15, 10, 5, 3]

plt.pie(cantidad, labels=mascotas, autopct="%1.1f%%", startangle=90)
plt.title("Mascotas favoritas de la clase")
plt.axis("equal")
plt.show()
```

## Ejercicio 5: Crecimiento de suscriptores (Gráfico de Líneas)

Dibuja un gráfico de líneas que represente el crecimiento de suscriptores de un canal de YouTube en el último
semestre.


```python
meses = ["Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"]
suscriptores = [1200, 1500, 2100, 3000, 4500, 6000]

plt.plot(meses, suscriptores, marker="o", color="red")
plt.title("Crecimiento de suscriptores (último semestre)")
plt.xlabel("Mes")
plt.ylabel("Suscriptores")
plt.grid(True)
plt.show()
```

##Ejercicio 6: Comparación de población en ciudades (Gráfico de Barras)

Crea un gráfico de barras para comparar la población de cuatro ciudades principales.


```python
ciudades = ["Bogotá", "Medellín", "Cali", "Barranquilla"]
poblacion = [8.0, 2.5, 2.2, 1.3]

plt.bar(ciudades, poblacion, color="steelblue")
plt.title("Población de ciudades principales")
plt.xlabel("Ciudad")
plt.ylabel("Población (millones)")
plt.show()
```

## Ejercicio 7: Gasto en publicidad vs. Ventas (Gráfico de Dispersión)

Representa mediante un diagrama de dispersión cómo el aumento del presupuesto en publicidad afecta el total
de ventas generadas.

```python
publicidad = [2, 4, 6, 8, 10, 12]
ventas = [10, 18, 25, 40, 48, 60]

plt.scatter(publicidad, ventas, color="purple")
plt.title("Gasto en publicidad vs. Ventas")
plt.xlabel("Presupuesto en publicidad")
plt.ylabel("Ventas")
plt.grid(True)
plt.show()
```

##Ejercicio 8: Presupuesto mensual del hogar (Gráfico de Pastel)
Elabora un gráfico circular que desglose en qué se gasta el dinero un hogar promedio al mes.


```python
categorias = ["Arriendo", "Alimentación", "Transporte", "Entretenimiento", "Ahorro"]
gastos = [40, 30, 15, 10, 5]

plt.pie(gastos, labels=categorias, autopct="%1.1f%%", startangle=90)
plt.title("Presupuesto mensual del hogar")
plt.axis("equal")
plt.show()
```

## Ejercicio 9: Variación de criptomonedas (Gráfico de Líneas)

Muestra la fluctuación del precio de una criptomoneda durante los últimos 5 días laborables.


```python
dias = ["Día 1", "Día 2", "Día 3", "Día 4", "Día 5"]
precio = [45000, 44200, 46800, 47100, 45900]

plt.plot(dias, precio, marker="o", color="goldenrod")
plt.title("Precio de la criptomoneda (últimos 5 días)")
plt.xlabel("Día")
plt.ylabel("Precio (USD)")
plt.grid(True)
plt.show()
```

##Ejercicio 10: Reto de Elección de Gráfico

**Escenario**: Un profesor quiere mostrar la distribución de las calificaciones finales de su grupo de estudiantes

divididas en rangos: Reprobado, Aprobado, Notable, Sobresaliente.
Analiza el caso y responde: ¿Qué tipo de gráfico de Matplotlib (plt.plot, plt.bar, plt.scatter o plt.pie) utilizarías y por
qué? Escribe el bloque de código con datos ficticios que resuelva este problema.


```python
rangos = ["Reprobado", "Aprobado", "Notable", "Sobresaliente"]
estudiantes = [4, 12, 9, 5]

plt.bar(rangos, estudiantes, color=["red", "orange", "skyblue", "green"])
plt.title("Distribución de calificaciones finales")
plt.xlabel("Rango de calificación")
plt.ylabel("Número de estudiantes")
plt.show()
```


Las calificaciones están agrupadas de reprobado a sobre saliente y podemos ver cuantos estudiantes estan en cada categoria
